```python
import pandas as pd

df = pd.read_excel("../data/raw/precio_vivienda_espana_2003_2025.xlsx")
df.head()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(11,6))

plt.plot(
    df["Año"],
    df["Precio general"],
    linewidth=2.5,
    marker="o",
    markersize=4
)

plt.title(
    "Evolución del precio medio de la vivienda libre en España (2003-2025)",
    fontsize=15,
    pad=15
)

plt.xlabel("Año", fontsize=12)
plt.ylabel("Precio (€/m²)", fontsize=12)

# Mostrar todos los años
plt.xticks(df["Año"], rotation=45)

# Cuadrícula horizontal únicamente
plt.grid(axis="y", alpha=0.3)

# Quitar margen innecesario
plt.margins(x=0.01)

plt.tight_layout()

# Guardar antes de mostrar
plt.savefig( "../figures/evolucion_precio_vivienda.png", dpi=300, bbox_inches="tight" )

plt.show()
```
